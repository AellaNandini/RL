```python



# VALUE ITERATION IN A DETERMINISTIC 2x2 GRID WORLD

# 1. Define parameters
gamma = 0.9
theta = 0.000001
goal_reward = 100
normal_reward = -1

states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

# 2. Define transition rules
# Format: (next_state, reward)
transitions = {
    'S0': {
        'Up': ('S0', normal_reward),
        'Down': ('S1', normal_reward),
        'Left': ('S0', normal_reward),
        'Right': ('S0', normal_reward)
    },

    'S1': {
        'Up': ('S0', normal_reward),
        'Down': ('S1', normal_reward),
        'Left': ('S1', normal_reward),
        'Right': ('G', goal_reward)
    }
}

# 3. Initialize state values
V = {
    'S0': 0.0,
    'S1': 0.0,
    'G': 0.0
}

# 4. Display transition information
print("STATE TRANSITION INFORMATION")
print("-" * 55)

for state in states:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, action, "->", next_state,
              "Reward:", reward)

# 5. Perform Value Iteration
iteration = 0
iteration_table = []

iteration_table.append(
    (iteration, V['S0'], V['S1'], 0.0)
)

while True:
    new_V = V.copy()
    delta = 0.0

    for state in states:
        action_values = []

        for action in actions:
            next_state, reward = transitions[state][action]

            q_value = reward + gamma * V[next_state]
            action_values.append(q_value)

        # Select maximum action value
        new_V[state] = max(action_values)

        # Calculate change in value
        change = abs(new_V[state] - V[state])
        delta = max(delta, change)

    # Update values simultaneously
    V = new_V
    iteration += 1

    iteration_table.append(
        (iteration, V['S0'], V['S1'], delta)
    )

    # Check convergence
    if delta < theta:
        break

# 6. Display iteration-wise state values
print("\nITERATION-WISE STATE VALUES")
print("-" * 65)
print("{:<12}{:<15}{:<15}{:<15}".format(
    "Iteration", "V(S0)", "V(S1)", "Delta"
))

for row in iteration_table:
    print("{:<12}{:<15.6f}{:<15.6f}{:<15.6f}".format(
        row[0], row[1], row[2], row[3]
    ))

# 7. Calculate final action values
print("\nFINAL ACTION VALUE TABLE")
print("-" * 55)
print("{:<10}{:<12}{:<12}{:<12}".format(
    "State", "Action", "Next State", "Q(s,a)"
))

policy = {}

for state in states:
    best_action = None
    best_value = float('-inf')

    for action in actions:
        next_state, reward = transitions[state][action]

        q_value = reward + gamma * V[next_state]

        print("{:<10}{:<12}{:<12}{:<12.2f}".format(
            state, action, next_state, q_value
        ))

        if q_value > best_value:
            best_value = q_value
            best_action = action

    policy[state] = best_action

# 8. Display final optimal state values
print("\nFINAL OPTIMAL STATE VALUES")
print("V*(S0) =", round(V['S0'], 2))
print("V*(S1) =", round(V['S1'], 2))
print("V*(G)  =", V['G'])

# 9. Display optimal policy
print("\nOPTIMAL POLICY")
for state in states:
    print(state, "->", policy[state])

print("\nNumber of iterations:", iteration)
print("Convergence achieved:", delta < theta)
print("Optimal path: S0 -> S1 -> G")
```
