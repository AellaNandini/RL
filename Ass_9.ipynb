```python
# Assignment 1: Stochastic Policy and Deterministic Environment

gamma = 0.9

# Part A: Define Bellman equations
# V(S0) = -1 + 0.81 * V(S1)
# V(S1) = 89.9 + 0.09 * V(S0)

# Part B: Solve equations algebraically

# Substitute V(S1) into V(S0)
V_S0 = 71.819 / 0.9271

# Calculate V(S1)
V_S1 = 89.9 + 0.09 * V_S0

print("STATE VALUES")
print("V(S0) =", round(V_S0, 2))
print("V(S1) =", round(V_S1, 2))

# Part C: Calculate action values

Q_S0_Down = -1 + gamma * V_S1
Q_S0_Right = -1

Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V_S0

print("\nACTION VALUES")
print("Q(S0, Down) =", round(Q_S0_Down, 2))
print("Q(S0, Right) =", round(Q_S0_Right, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))
print("Q(S1, Up) =", round(Q_S1_Up, 2))

# Part D: Verify Bellman expectation equations

check_S0 = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
check_S1 = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

print("\nBELLMAN VERIFICATION")
print("V(S0) =", round(check_S0, 2))
print("V(S1) =", round(check_S1, 2))

print("\nProgram completed successfully!")
```

```python
# Assignment 2: Stochastic Environment and Deterministic Policy

gamma = 0.9
theta = 0.000001

# Initial state values
V_S0 = 0.0
V_S1 = 0.0

iteration = 0

# Part A: Iterative Policy Evaluation

while True:

    # Store previous values
    old_S0 = V_S0
    old_S1 = V_S1

    # Calculate new values using previous values
    new_S0 = 0.1 * (-1 + gamma * 0) + \
             0.9 * (-1 + gamma * old_S1)

    new_S1 = 0.9 * (100 + gamma * 0) + \
             0.1 * (-1 + gamma * old_S0)

    # Calculate maximum change
    delta = max(
        abs(new_S0 - old_S0),
        abs(new_S1 - old_S1)
    )

    # Update state values
    V_S0 = new_S0
    V_S1 = new_S1

    iteration += 1

    # Stop when converged
    if delta < theta:
        break

# Part B: Display state values

print("ITERATIVE POLICY EVALUATION")
print("Number of iterations:", iteration)

print("V(S0) =", round(V_S0, 2))
print("V(S1) =", round(V_S1, 2))

# Part C: Calculate action values

# S0 Down: Wall (0.1), S1 (0.9)
Q_S0_Down = 0.1 * (-1 + gamma * 0) + \
            0.9 * (-1 + gamma * V_S1)

# S0 Right: Wall
Q_S0_Right = -1 + gamma * 0

# S1 Right: Goal (0.9), S0 (0.1)
Q_S1_Right = 0.9 * (100 + gamma * 0) + \
             0.1 * (-1 + gamma * V_S0)

# S1 Up: S0
Q_S1_Up = -1 + gamma * V_S0

print("\nACTION VALUES")
print("Q(S0, Down) =", round(Q_S0_Down, 2))
print("Q(S0, Right) =", round(Q_S0_Right, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))
print("Q(S1, Up) =", round(Q_S1_Up, 2))

# Part D: Verify policy consistency

print("\nPOLICY VERIFICATION")

print("V(S0) =", round(V_S0, 2),
      "Q(S0, Down) =", round(Q_S0_Down, 2))

print("V(S1) =", round(V_S1, 2),
      "Q(S1, Right) =", round(Q_S1_Right, 2))

if abs(V_S0 - Q_S0_Down) < theta and \
   abs(V_S1 - Q_S1_Right) < theta:

    print("Policy consistency verified successfully!")

else:
    print("Verification failed!")

print("\nProgram completed successfully!")
```
