```python

import random

# -----------------------------
# Part A - Machine Initialization
# -----------------------------
machines = ['A', 'B', 'C']
epsilon = 0.2
N_play = 10

# Initialize tables
N = {m: 0 for m in machines}      # Number of plays
Q = {m: 0.0 for m in machines}    # Estimated Q-values
Rewards = {m: [] for m in machines}  # Reward history

total_reward = 0

print("ε-Greedy Multi-Armed Bandit")
print("Machines:", machines)
print("Epsilon =", epsilon)
print("-" * 50)

# -----------------------------
# Initial Knowledge (Play each machine once)
# -----------------------------
print("\nInitial Testing (Each machine once)\n")

for m in machines:
    reward = int(input(f"Enter reward for Machine {m}: "))
    Rewards[m].append(reward)
    N[m] += 1
    Q[m] = reward
    total_reward += reward

print("\nInitial Q Values")
for m in machines:
    print(m, "N =", N[m], "Q =", round(Q[m],2))

# -----------------------------
# Remaining Plays
# -----------------------------
for play in range(4, N_play + 1):

    print("\n==========================")
    print("Play", play)

    r = random.random()
    print("Random Number =", round(r,2))

    # Exploration
    if r < epsilon:
        action = random.choice(machines)
        print("Decision : Exploration")
    else:
        max_q = max(Q.values())
        best = [m for m in machines if Q[m] == max_q]
        action = random.choice(best)
        print("Decision : Exploitation")

    print("Selected Machine :", action)

    reward = int(input("Enter Reward: "))

    total_reward += reward
    Rewards[action].append(reward)

    old_q = Q[action]
    N[action] += 1

    # Incremental Sample Average
    Q[action] = old_q + (reward - old_q) / N[action]

    print("Updated Q(", action, ") =", round(Q[action],2))

# -----------------------------
# Final Tables
# -----------------------------
print("\n==============================")
print("FINAL N TABLE")
for m in machines:
    print(m, ":", N[m])

print("\nFINAL REWARD TABLE")
for m in machines:
    print(m, ":", Rewards[m])

print("\nFINAL Q TABLE")
for m in machines:
    print(m, ":", round(Q[m],2))

print("\nTotal Reward =", total_reward)
print("Average Reward =", round(total_reward / N_play,2))
```
